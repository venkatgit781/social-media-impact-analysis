```python
df1["Weekly_Usage_Hours"] = df1["Daily_Usage_Hours"] * 7 + df1["Weekend_Extra_Hours"] * 2
 
df1["Usage_Category"] = pd.cut(
    df1["Daily_Usage_Hours"],
    bins=[0, 3, 6, 9, np.inf],
    labels=["Low (0-3h)", "Moderate (3-6h)", "High (6-9h)", "Very High (9h+)"],
)
 
df1["Enough_Sleep"] = np.where(df1["Sleep_Duration_Hours"] >= 7, "Yes (7h+)", "No (<7h)")

```

```python
# Group by platform and sort
print("\nPlatforms ranked by average daily usage:")
platform_summary = (
        df1.groupby("Primary_Platform")
        .agg(
            students=("Student_ID", "count"),
            avg_usage=("Daily_Usage_Hours", "mean"),
            avg_gpa=("Academic_Performance_GPA", "mean"),
            avg_mental_health=("Mental_Health_Index", "mean"),
        )
        .sort_values("avg_usage", ascending=False)
        .round(2)
    )
print(platform_summary)
```

```python
# Pivot table: rows = academic level, columns = usage category, values = average GPA
print("\nAverage GPA by academic level and usage category (pivot table):")
pivot = pd.pivot_table(
        df1, values="Academic_Performance_GPA", index="Academic_Level",
        columns="Usage_Category", aggfunc="mean", observed=False,
    ).round(2)
print(pivot)
```

```python
# Crosstab: percentage of each impact type within each comparison frequency
COMPARISON_ORDER = ["Never", "Rarely", "Sometimes", "Frequently", "Always"]
print("\nOverall impact (%) by how often students compare themselves to others:")
cross = pd.crosstab(df1["Social_Comparison_Frequency"], df1["Overall_Impact"], normalize="index") * 100
print(cross.reindex(index=COMPARISON_ORDER, columns=IMPACT_ORDER).round(1))
```

```python
# Filtering rows with conditions
at_risk = df1[(df1["Daily_Usage_Hours"] > 8) & (df1["Late_Night_Usage"]) & (df1["Sleep_Duration_Hours"] < 6)]
print(f"\n'At-risk' students (>8h usage, late-night use, <6h sleep): {len(at_risk)}")
print(f"Their average GPA: {at_risk['Academic_Performance_GPA'].mean():.2f} "
f"vs everyone: {df1['Academic_Performance_GPA'].mean():.2f}")
```

```python
# Late-night usage effect
print("\nAverage sleep and mental health: late-night users vs others")
print(df1.groupby("Late_Night_Usage")[["Sleep_Duration_Hours", "Mental_Health_Index"]].mean().round(2))
```

```python
#  Histogram - how is daily usage spread out?
plt.figure(figsize=(8, 5))
plt.hist(df1["Daily_Usage_Hours"], bins=30, color="#4C72B0", edgecolor="white")
plt.axvline(df1["Daily_Usage_Hours"].mean(), color="red", linestyle="--",
                label=f"Average = {df1['Daily_Usage_Hours'].mean():.1f}h")
plt.title("How Many Hours Do Students Spend on Social Media Each Day?")
plt.xlabel("Daily usage (hours)")
plt.ylabel("Number of students")
plt.legend()
plt.savefig('my_plot.png')
```

```python
# Bar chart - most popular platforms
counts = df1["Primary_Platform"].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(counts.index, counts.values, color="#55A868")
plt.title("Most Popular Social Media Platforms")
plt.xlabel("Platform")
plt.ylabel("Number of students")
plt.xticks(rotation=30)
plt.savefig('my_barchart.png')
```

```python
# Pie chart - overall impact
IMPACT_COLORS = {"Beneficial": "#55A868", "Neutral": "#CCB974", "Negative": "#C44E52"}
impact_counts = df1["Overall_Impact"].value_counts().reindex(IMPACT_ORDER)
plt.figure(figsize=(6, 6))
plt.pie(impact_counts, labels=impact_counts.index, autopct="%1.1f%%", startangle=90,
        colors=[IMPACT_COLORS[name] for name in impact_counts.index],
        explode=[0, 0.05, 0.12], pctdistance=0.75)
plt.title("Overall Impact of Social Media on Students")
plt.savefig('my_piechart.png')
```

```python
# Scatter plot - usage vs mental health, with a trend line from NumPy
x = df1["Daily_Usage_Hours"]
y = df1["Mental_Health_Index"]
slope, intercept = np.polyfit(x, y, 1)  # best straight line through the points
plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.3, s=10, color="#8172B2")
xs = np.linspace(x.min(), x.max(), 100)
plt.plot(xs, slope * xs + intercept, color="red", linewidth=2, label="Trend line")
plt.title("More Social Media Time, Lower Mental Health Score")
plt.xlabel("Daily usage (hours)")
plt.ylabel("Mental Health Index (higher = better)")
plt.legend()
plt.savefig('my_scatterplot.png')
```

```python
# Correlation heatmap - which numbers move together?
numeric_cols = ["Age", "Daily_Usage_Hours", "Weekend_Extra_Hours", "Sleep_Duration_Hours",
                "Sleep_Quality_Score", "Perceived_Stress_Score", "Mental_Health_Index",
                "Academic_Performance_GPA"]
plt.figure(figsize=(9, 7))
sns.heatmap(df1[numeric_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
            square=True, linewidths=0.5)
plt.title("Correlation Heatmap\n(red = move together, blue = move in opposite directions)")
plt.savefig('my_heatmap.png')
```

```python
# Box plot - GPA for each impact group
plt.figure(figsize=(8, 5))
sns.boxplot(data=df1, x="Overall_Impact", y="Academic_Performance_GPA", order=IMPACT_ORDER,
            hue="Overall_Impact", palette=IMPACT_COLORS, legend=False)
plt.title("GPA by Overall Social Media Impact")
plt.xlabel("Overall impact")
plt.ylabel("GPA")
plt.savefig('my_boxplot.png')
```

```python
# Violin plot - sleep hours for late-night users vs others
plt.figure(figsize=(8, 5))
late_night_labels = df1["Late_Night_Usage"].map({True: "Yes", False: "No"})
sns.violinplot(data=df1, x=late_night_labels, y="Sleep_Duration_Hours", order=["No", "Yes"],
                hue=late_night_labels, hue_order=["No", "Yes"], legend=False)
plt.title("Sleep Duration: Late-Night Users vs Others")
plt.xlabel("Uses social media late at night?")
plt.ylabel("Sleep (hours)")
plt.savefig('my_violinplot.png')
```

```python
# Count plot - social comparison vs impact
plt.figure(figsize=(9, 5))
sns.countplot(data=df1, x="Social_Comparison_Frequency", hue="Overall_Impact",
                  order=COMPARISON_ORDER, hue_order=IMPACT_ORDER, palette=IMPACT_COLORS)
plt.title("How Often Students Compare Themselves to Others, by Impact")
plt.xlabel("Social comparison frequency")
plt.ylabel("Number of students")
plt.savefig('my_countplot.png')
```

```python
# Bar plot with averages - stress by usage category
plt.figure(figsize=(8, 5))
sns.barplot(data=df1, x="Usage_Category", y="Perceived_Stress_Score", hue="Usage_Category",
            legend=False, errorbar=None, dodge=False)
plt.title("Average Stress Score by Usage Category")
plt.xlabel("Daily usage category")
plt.ylabel("Average stress score")
plt.savefig('my_barplot.png')
```

```python
# Pair plot - many scatter plots at once (uses a sample of 800 rows to stay fast)
sample = df1.sample(n=800, random_state=42)
grid = sns.pairplot(sample, vars=["Daily_Usage_Hours", "Sleep_Duration_Hours",
                                  "Mental_Health_Index", "Academic_Performance_GPA"],
                    hue="Overall_Impact", hue_order=IMPACT_ORDER, palette=IMPACT_COLORS, plot_kws={"alpha": 0.5, "s": 15})
grid.figure.suptitle("Pair Plot of Key Measures", y=1.02)
plt.savefig('my_pairplot.png')
```
